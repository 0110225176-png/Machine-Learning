```python
from sklearn.model_selection import train_test_split

# Menentukan X dan y
X = df.loc[:, 'season':'registered']
y = df['cnt']

# Membagi 80% training dan 20% testing
X_train_temp, X_test, y_train_temp, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Mengambil 10% dari data training sementara
# sebagai validation
X_train, X_val, y_train, y_val = train_test_split(
    X_train_temp,
    y_train_temp,
    test_size=0.10,
    random_state=42
)

# Menampilkan jumlah data
print("Jumlah Data Training   :", len(X_train))
print("Jumlah Data Validation :", len(X_val))
print("Jumlah Data Testing    :", len(X_test))
```

```python
print("===== DATA TRAINING =====")
print("Jumlah data:", len(X_train))
display(pd.concat([X_train, y_train], axis=1).head())

print("\n===== DATA VALIDATION =====")
print("Jumlah data:", len(X_val))
display(pd.concat([X_val, y_val], axis=1).head())

print("\n===== DATA TESTING =====")
print("Jumlah data:", len(X_test))
display(pd.concat([X_test, y_test], axis=1).head())
```
